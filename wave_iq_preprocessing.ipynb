```python
id_analysis = (
    df.groupby("ID")
      .apply(lambda group: pd.Series({
          "Total_Rows": len(group),
          "Unique_Records": len(group.drop_duplicates())
      }), include_groups=False)
      .reset_index()
)

id_analysis["Repeated_Rows"] = (
    id_analysis["Total_Rows"] - id_analysis["Unique_Records"]
)

display(
    id_analysis.sort_values(
        "Total_Rows", ascending=False
    ).head(10)
)
```
