```python
# ==========================================
# INVESTIGATION OF SUSPICIOUS DATA VALUES
# ==========================================

# 1. Find non-numeric values in act_shrink%
shrink_numeric = pd.to_numeric(
    df_check["act_shrink%"], errors="coerce"
)

invalid_shrink = df_check[
    shrink_numeric.isna()
]

print("NON-NUMERIC act_shrink% VALUES")
print("Total records:", len(invalid_shrink))

display(
    invalid_shrink["act_shrink%"]
    .value_counts(dropna=False)
    .head(20)
)

# Show some example records
display(
    invalid_shrink[
        ["ID", "Previous_pdn", "act_shrink%", "Rej_and_cut_Piece"]
    ].head(15)
)


# 2. Check whether non-numeric shrink values occur
# in TOTAL rows or ordinary rows
print("\nSHRINKAGE VALUES BY RECORD TYPE")

invalid_shrink_mask = shrink_numeric.isna()

total_mask = (
    df_check["Previous_pdn"]
    .astype(str).str.strip().str.upper().eq("TOTAL")
)

print("Invalid shrink values in TOTAL rows:",
      (invalid_shrink_mask & total_mask).sum())

print("Invalid shrink values in non-TOTAL rows:",
      (invalid_shrink_mask & ~total_mask).sum())


# 3. Investigate all unusual weft_count values
unusual_weft = df_check[df_check["weft_count"] > 200]

print("\nUNUSUAL WEFT COUNT FREQUENCIES")

display(
    unusual_weft["weft_count"]
    .value_counts()
    .sort_index()
)

print("\nCONSTRUCTION PATTERNS OF UNUSUAL RECORDS")

display(
    unusual_weft[
        ["ID", "Construction", "warp_count",
         "weft_count", "epi", "ppi", "Previous_pdn"]
    ].drop_duplicates().head(30)
)


# 4. Check the distribution of TOTAL and non-TOTAL rows
print("\nRECORD TYPE COUNTS")

print("TOTAL rows:", total_mask.sum())
print("Non-TOTAL rows:", (~total_mask).sum())

print("\nTARGET SUMMARY BY RECORD TYPE")

display(
    df_check.assign(
        Record_Type=total_mask.map({
            True: "TOTAL",
            False: "Non-TOTAL"
        })
    )
    .groupby("Record_Type")["Rej_and_cut_Piece"]
    .describe()
)
```

```python
# ==========================================
# CHECK WHETHER TOTAL ROWS SUM INDIVIDUAL ROWS
# ==========================================

df_test = df_check.copy()

# Identify TOTAL rows
is_total = (
    df_test["Previous_pdn"]
    .astype(str).str.strip().str.upper().eq("TOTAL")
)

# Separate the two record types
detail_rows = df_test[~is_total].copy()
total_rows = df_test[is_total].copy()

# Calculate rejection totals for individual records
detail_summary = (
    detail_rows.groupby("ID")
    .agg(
        Detail_Record_Count=("ID", "size"),
        Detail_Rejection_Sum=("Rej_and_cut_Piece", "sum")
    )
)

# Calculate recorded TOTAL rejection values
order_summary = (
    total_rows.groupby("ID")
    .agg(
        Total_Record_Count=("ID", "size"),
        Recorded_Total_Rejection=("Rej_and_cut_Piece", "sum")
    )
)

# Compare both
comparison = detail_summary.join(order_summary, how="inner")

comparison["Difference"] = (
    comparison["Recorded_Total_Rejection"]
    - comparison["Detail_Rejection_Sum"]
)

print("IDs containing both record types:", len(comparison))

print("\nComparison results:")
display(comparison.head(20))

print("\nExact matches:",
      (comparison["Difference"] == 0).sum())

print("Non-matches:",
      (comparison["Difference"] != 0).sum())

print("\nDifference statistics:")
display(comparison["Difference"].describe())
```

```python

# Create a working copy of the original dataset
df_work = df_check.copy()

# Identify order-level TOTAL rows
is_total = (
    df_work["Previous_pdn"]
    .astype(str)
    .str.strip()
    .str.upper()
    .eq("TOTAL")
)

# Keep individual production/beam records only
df_detail = df_work[~is_total].copy()

# Display the results
print("Original dataset shape:", df_work.shape)
print("TOTAL summary rows removed:", is_total.sum())
print("Individual production/beam records:", df_detail.shape)

# Verify that no TOTAL rows remain
remaining_total = (
    df_detail["Previous_pdn"]
    .astype(str)
    .str.strip()
    .str.upper()
    .eq("TOTAL")
)

print("Remaining TOTAL rows:", remaining_total.sum())
```

```python

# Number of records if exact duplicates are counted once
df_unique_check = df_detail.drop_duplicates().copy()

print("Original individual records:", len(df_detail))
print("Unique records:", len(df_unique_check))
print("Duplicate entries:", len(df_detail) - len(df_unique_check))

# How many unique records have duplicates?
duplicate_groups = (
    df_detail[df_detail.duplicated(keep=False)]
    .drop_duplicates()
)

print("Unique records belonging to duplicate groups:", len(duplicate_groups))

# Find the most frequently repeated exact records
duplicate_frequency = (
    df_detail.value_counts(dropna=False)
    .reset_index(name="Frequency")
)

duplicate_frequency = duplicate_frequency[
    duplicate_frequency["Frequency"] > 1
].sort_values("Frequency", ascending=False)

print("\nTop 15 most frequently repeated records:")
display(duplicate_frequency.head(15))
```

```python

# Count records and unique records for each ID
id_duplicate_summary = (
    df_detail.groupby("ID", dropna=False)
    .agg(
        Total_Rows=("ID", "size"),
        Unique_Records=("ID", "nunique")
    )
)

# Count exact duplicate entries for each ID
id_duplicate_summary["Duplicate_Entries"] = (
    df_detail.groupby("ID", dropna=False)
    .apply(lambda group: group.duplicated().sum(),
           include_groups=False)
)

id_duplicate_summary = id_duplicate_summary.sort_values(
    "Duplicate_Entries", ascending=False
)

display(id_duplicate_summary.head(20))
```

```python

# Calculate the number of rows and distinct complete records per ID
id_summary = (
    df_detail.groupby("ID", dropna=False)
    .apply(
        lambda group: pd.Series({
            "Total_Rows": len(group),
            "Distinct_Records": len(group.drop_duplicates()),
            "Duplicate_Entries": group.duplicated().sum()
        }),
        include_groups=False
    )
    .sort_values("Duplicate_Entries", ascending=False)
)

print("Number of distinct IDs:", df_detail["ID"].nunique())
print("\nTop 20 IDs with the most duplicate entries:")

display(id_summary.head(20))
```
